本篇文章详细介绍了ModelScope创空间的创建与搭建过程。

>  **特别说明：** 您可以自由创建多个创空间并分享给您的朋友，但考虑到首页的空间展示质量，您所创建的空间并不会自动进入创空间首页精选列表，如您有意愿申请进入首页列表，非常欢迎与我们取得联系（钉钉群：8010015744，邮箱：contact@modelscope.cn）

## 创建自己的创空间
要创建文件并共享至ModelScope社区，您需要拥有ModelScope账号。如果您还没有账号，请注册一个新账号以用于向Model hub验证您的身份。

![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666332718781-f1a5b454-e9b3-4c5e-94b2-b7847c7e6484.png)

您可以通过ModelScope的web页面创建创空间，注册登录后，在头像左侧会出现创建按钮，点击可创建模型、数据集、创空间。点击创建，进入创空间页面填写基础信息实现创空间的创建。
![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666333002866-9acc2d21-3e64-4e7c-a85c-de5c92dd0844.png)

创建空间分为两个步骤： 

- (1）填写基础信息。基础信息包括您的模型英文名称、中文名、所有者、许可证类型、是否公开和创空间描述。

  - 许可证类型决定您的创空间遵循对应的开源协议。
  - 是否公开决定您的创空间是否能被其他用户检索查看，若设置为非公开创空间，则其他用户无法查看，仅您自己查看。您也可以创建后在设置页面进行权限的修改配置。
  - 创空间描述建议介绍您的空间的特性和应用场景，将展现在空间列表页方便用户搜索查询。

  - ![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666691532199-52be71d9-3303-4825-bb33-2be32a45adb1.png)
- (2）完成创建后，平台将为您分配一个git存储地址，可在创空间详情页中获取，如下： 
  - ![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666332719498-d25715f6-8ec0-4fb9-98c7-4baab96860fd.png) 


您可通过页面或者git的方式将文件添加至该创空间库中，也可通过页面上传相关文件：

- (1）使用git方式添加创空间文件
```yaml
git lfs install
  #根据实际情况填写
git clone http://oauth2:<your_access_token>@www.modelscope.cn/studios/<your_studio_path>/<your_studio_name>.git

  #创建 app.py后添加到git仓库
git add app.py
git commit -m "add app.py"
git push
```
>  其中ACCESS_TOKEN可以通过用账号密码登录网站, 前往【个人中心】->【访问令牌】获取：
 ![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666691636402-f41d117f-34c4-4f68-8767-b1a6222258a7.png)


- (2）使用控制台添加创空间文件

 - ![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1667311588859-26e91e25-119a-41cf-b5fa-5f3a2d23858e.png)


# 创空间开发
登录ModelScope网站，在已登录的用户账号名下点击[个人中心](https://pre.modelscope.cn/my/myspace) ，选择“我的创空间”，展示当前用户下所有的创空间列表
![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666339801424-b5b5175e-bca6-4740-a790-b1287a3af978.png)

点击空间详情，可以查看空间展示效果、空间文件列表、空间设置
其中空间文件列表展示空间当中的开发代码文件、配置、图片、视频等信息，上传空间文件的方式可参考创空间介绍
![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666662949481-13d98452-c721-426c-ac5c-73507428f302.png)

README.md是每个创空间必须包含的文件，具体含义见创空间卡片章节。
创空间需指定启动脚本，默认为根目录下的app.py文件，具体使用方式见创空间卡片README.md entry_file说明。
## 依赖安装
创空间支持用户自定义安装三方依赖，使用方法，在根目录下新建requirements.txt文件存放依赖列表即可，例如：



````
wget
openssl
````


创空间默认已安装notebook环境下所有依赖，用户无需额外安装，并且创空间默认安装了gradio，版本号为3.4.0，后续更多版本待开放

## 创空间重启
创空间代码更新后，可直接点击重启空间展示按钮进行重部署，系统将会拉取最新的空间文件进行空间部署，部署后将在“空间展示”一栏显示最新的空间样式
![image.png](https://resouces.modelscope.cn/document/docdata/2025-4-29_13:51/dist/%E5%88%9B%E7%A9%BA%E9%97%B4/resources/1666669086614-59bf6ecd-f785-45ed-b633-2d89f4ae371c.png)
